```python
#cleaning of the data
#check null values
df.isna().sum()
```

```python
#handle the null values 
#for retailer fill with Not Provided or Unkwon
#handle all the strings columns null values by filling with unknown
strings=[
    'Retailer',
    'Region',
    'State',
    'City',
    'Product',
    'Sales Method'
]

df[strings].fillna('Unknown',inplace=True)
```

```python
#handle the date column
#check the data type of the date column
df['Invoice Date'].dtype
```

```python
#check for dates missing values
df['Invoice Date'].isna().sum()
```

```python
#look for unique entries
df['Invoice Date'].unique()
```

```python
#cleaning the date column
#removing the white space
df['Invoice Date']=df['Invoice Date'].str.strip()
```

```python
#make uniform the null values
df['Invoice Date'].replace(
    ['Unkwon','N/A','','Not Available'],
    pd.NA,
    inplace=True
)
    
```

```python
#converting to datetime 
df['Invoice Date']=pd.to_datetime(
    df['Invoice Date'],
    errors='coerce'
)
                                  
```

```python
#check failed dates
df[df['Invoice Date'].isna()]
```

```python
df['Invoice Date'].dtype
```

```python
#extracting required components
df['Year_Num']=df['Invoice Date'].dt.year
df['Month_Num']=df['Invoice Date'].dt.month
df['Month_Name']=df['Invoice Date'].dt.month_name()
df['Day_Num']=df['Invoice Date'].dt.day
df['Day_Name']=df['Invoice Date'].dt.day_name()
```

```python
#HANDLING currency/numeric columns columns (Price per Unit,Units Sold,Total Sales,Operating Profit,Operating Margin)
#check the data type for the column
currency_columns = [
    'Total Sales',
    'Operating Profit',
    'Price per Unit',
    'Units Sold'
]
```

```python
#check the data types
df[currency_columns].dtypes
```

```python
#check the missing values
df[currency_columns].isna().sum()
```

```python
#remove leading and trailing spaces
for col in currency_columns:
    df[col] = df[col].str.strip()
```

```python
#replace invalid entries
for col in currency_columns:
    df[col] = df[col].replace(
        ['Unknown', 'N/A', '', '-'],
        pd.NA
    )
```

```python
#remove currency symbols
for col in currency_columns:
    df[col] = df[col].str.replace('$', '', regex=False)
    df[col] = df[col].str.replace('KES', '', regex=False)
    df[col] = df[col].str.replace('USD', '', regex=False)
    df[col] = df[col].str.replace('€', '', regex=False)
    df[col] = df[col].str.replace('£', '', regex=False)
```

```python
#remove commas
for col in currency_columns:
    df[col] = df[col].str.replace(',', '', regex=False)
```

```python
#convert every column to numeric
for col in currency_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors='coerce'
    )
```

```python
df[currency_columns].dtypes
```

```python
#handling the % columns
df['Operating Margin'].head
```

```python
#check the data types
df['Operating Margin'].dtype
```

```python
#removing the % symbol
df['Operating Margin'] = df['Operating Margin'].str.replace('%', '', regex=False)
```

```python
#converting to numeric values
df['Operating Margin'] = pd.to_numeric(
        df['Operating Margin'],
        errors='coerce'
    )
```

```python
#converting to decimal form
df['Operating Margin']=df['Operating Margin']/100
```

```python
#cleaning the column names
df.columns=(
    df.columns
    .str.strip()
    .str.replace(' ','_')
    .str.title()
)
```
